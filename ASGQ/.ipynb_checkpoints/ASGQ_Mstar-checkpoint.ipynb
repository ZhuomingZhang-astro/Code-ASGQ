```python
import numpy as np


def Calculate_Ratio_And_Check(sgaldata, cgal_data, ratio_min):
    """Calculates target ratios, verifies if any ratio falls below ratio_min,

    and returns the minimum valid ratio value.

    Parameters:
    -----------
    sgaldata : numpy.ndarray
        Array containing satellite galaxy data
        Key columns: index 3 (numerator) and index 9 (match key).
    cgal_data : numpy.ndarray
        Array containing central galaxy data
        Key columns: index 8 (match key) and index 18 (denominator).
    ratio_min : float
        Minimum allowed ratio threshold for validation.

    Returns:
    --------
    float
        The minimum ratio value among all calculated valid ratios.

    Raises:
    -------
    ValueError
        If a ratio falls below ratio_min, a denominator equals zero, or no
        match is found.
    """
    # List to store all valid calculated ratios
    all_ratios = []

    # Iterate through each row in sgaldata
    for i in range(sgaldata.shape[0]):
        # 1. Extract target values from current row
        numerator = sgaldata[i, 3]  # Numerator value
        match_value = sgaldata[i, 9]  # Match criterion key for cgal_data[:, 8]

        # 2. Find row indices in cgal_data where column 8 matches match_value
        match_indices = np.where(cgal_data[:, 8] == match_value)[0]

        # Check if matching central galaxy data exists
        if len(match_indices) == 0:
            raise ValueError(
                f"sgaldata row {i}: No match found in cgal_data for cgal_data[:, 8] = {match_value}"
            )

        # 3. Iterate through matching rows, compute ratio and validate
        for k in match_indices:
            denominator = cgal_data[k, 18]  # Denominator value

            # Avoid division by zero
            if denominator == 0:
                raise ValueError(
                    f"sgaldata row {i} matched with cgal_data row {k}: "
                    f"Denominator cgal_data[{k}, 18] = 0, division by zero."
                )

            # Compute ratio
            ratio = numerator / denominator
            all_ratios.append(ratio)

            # Check if ratio falls below threshold
            if ratio < ratio_min:
                raise ValueError(
                    f"Ratio is below ratio_min! Details:\n"
                    f"sgaldata row {i}: sgaldata[{i}, 3] = {numerator}\n"
                    f"cgal_data row {k}: cgal_data[{k}, 8] = {match_value} (matched key), "
                    f"cgal_data[{k}, 18] = {denominator}\n"
                    f"Calculated ratio: {numerator} / {denominator} = {ratio:.4f} < {ratio_min}"
                )

    # 4. Check for non-empty ratio list and return minimum value
    if len(all_ratios) == 0:
        raise ValueError("No valid ratios calculated. Please check input data.")

    min_ratio = np.min(all_ratios)
    print(f"Minimum valid ratio: {min_ratio:.4f}")
    return min_ratio


def Mass_Function(mass_array, bins):
    """Calculates unnormalized mass function, automatically filtering out non-

    positive mass values.

    Parameters:
    -----------
    mass_array : numpy.ndarray
        Array containing galaxy stellar/halo mass values.
    bins : array-like
        Edges for log-mass binning.

    Returns:
    --------
    mass_mid : numpy.ndarray
        Median log-mass value for each bin.
    log_counts : numpy.ndarray
        Log10 of normalized frequency density per bin.
    edges : numpy.ndarray
        Bin edges.
    """
    # Filter out non-positive mass values
    mass_valid = mass_array[mass_array > 0]
    log_mass = np.log10(mass_valid)

    hist, edges = np.histogram(log_mass, bins=bins, density=True)

    # Calculate median mass for samples within each bin
    bin_medians = []
    for i in range(len(edges) - 1):
        # Mask samples within current bin
        mask = (log_mass >= edges[i]) & (log_mass < edges[i + 1])
        bin_samples = log_mass[mask]

        # Calculate median (fallback to bin center if bin is empty)
        if len(bin_samples) > 0:
            bin_medians.append(np.median(bin_samples))
        else:
            bin_medians.append((edges[i] + edges[i + 1]) / 2)

    mass_mid = np.array(bin_medians)

    # Compute log10 of histogram counts, suppressing division warnings
    with np.errstate(divide="ignore"):
        log_counts = np.log10(hist)

    return mass_mid, log_counts, edges


def Classify_Galaxies_By_SFR_Mstar(star_mass, sfr, redshift=0.0):
    """Classifies galaxies into Star-Forming (SF) and Quenched (Q) populations

    based on a redshift-dependent Star Formation Rate (SFR) threshold.

    Parameters:
    -----------
    star_mass : numpy.ndarray
        Stellar mass values.
    sfr : numpy.ndarray
        Star Formation Rate values.
    redshift : float, optional
        Redshift of the snapshot (default is 0.0).

    Returns:
    --------
    m_SF : numpy.ndarray
        Stellar masses of star-forming galaxies.
    m_GV : numpy.ndarray
        Stellar masses of green valley galaxies (currently empty by design).
    m_Q : numpy.ndarray
        Stellar masses of quenched galaxies.
    """
    SF = []
    GV = []
    Q = []

    for m, s in zip(star_mass, sfr):
        if m <= 0:
            continue

        log_m = np.log10(m)
        log_s = np.log10(s) if s > 0 else -np.inf

        # Calculate redshift-dependent SFR threshold
        threshold = 0.73 * log_m - 7.33 + np.log10((1 + redshift) ** 2)
        threshold_q = threshold - 1

        # Classification logic based on SFR threshold
        if s > 0 and log_s > threshold:
            SF.append((m, s))
        elif s > 0 and threshold_q < log_s < threshold:
            SF.append((m, s))
        else:
            Q.append((m, s))

    m_SF = np.array([x[0] for x in SF]) if len(SF) > 0 else np.array([])
    m_GV = np.array([x[0] for x in GV]) if len(GV) > 0 else np.array([])
    m_Q = np.array([x[0] for x in Q]) if len(Q) > 0 else np.array([])

    return m_SF, m_GV, m_Q


def Calc_Stats_For_Errorbar(y_list, percentile_range=[0.0015, 0.5, 0.9985]):
    """Calculates median and asymmetric error bars corresponding to a 99.7%

    (3σ) confidence interval.

    Parameters:
    -----------
    y_list : list or numpy.ndarray
        Input data array of shape (N_samples, N_bins).
    percentile_range : list, optional
        Percentiles for [lower limit, median, upper limit].
        Default [0.0015, 0.5, 0.9985] covers 99.7% of samples (3σ).

    Returns:
    --------
    y_median : numpy.ndarray
        Median values along axis 0.
    y_err : list of numpy.ndarray
        List containing [lower_errors, upper_errors] relative to median.
    """
    y_array = np.array(y_list)  # shape: (N_samples, N_bins)
    y_median = np.percentile(y_array, 50, axis=0)
    y_low = np.percentile(y_array, percentile_range[0] * 100, axis=0)
    y_high = np.percentile(y_array, percentile_range[2] * 100, axis=0)

    # Asymmetric error bounds calculation
    y_err = [y_median - y_low, y_high - y_median]
    return y_median, y_err


def Calc_Quenched_Fraction(mass_total, mass_quenched, bins):
    """Calculates the quenched fraction of galaxies across specified mass bins.

    Parameters:
    -----------
    mass_total : numpy.ndarray
        Stellar masses of all galaxies.
    mass_quenched : numpy.ndarray
        Stellar masses of quenched galaxies.
    bins : array-like
        Log-mass bin edges.

    Returns:
    --------
    q_frac : numpy.ndarray
        Fraction of quenched galaxies in each bin.
    """
    # Count total and quenched galaxies within identical mass bins
    hist_total, _ = np.histogram(np.log10(mass_total), bins=bins)
    hist_q, _ = np.histogram(np.log10(mass_quenched), bins=bins)

    # Prevent division by zero for empty bins
    hist_total[hist_total == 0] = 1
    q_frac = hist_q / hist_total

    return q_frac
```

```python
# ===================== 1. Simulation Configurations =====================
simulations_config = [
    {
        "name": "SIMBA",
        "snapnum_center": "151",
        "data_base_dir": "/data/inspur_disk06/userdir/zhangzhuoming/Quench_Direction_Data/SIMBAm100n1024/data/",
        "starmass_min": 1.82 * 1e9,
        "redshift": 0.0,
        "starmass_bins": [
            np.log10(1.82 * 1e9 * 10**x) for x in [0, 0.5, 1, 10]
        ],
    },
    {
        "name": "TNG100",
        "snapnum_center": "099",  # Assuming z=0 snapshot is 099 for TNG100
        "data_base_dir": "/data/inspur_disk06/userdir/zhangzhuoming/Quench_Direction_Data/TNG100/data/",
        "starmass_min": 1.40 * 1e8,
        "redshift": 0.0,
        "starmass_bins": [
            np.log10(1.40 * 1e8 * 10**x) for x in [0, 0.5, 1, 1.5, 2, 10]
        ],
    },
    {
        "name": "EAGLE",
        "snapnum_center": "028_z000p000",  # Assuming z=0 snapshot is 028 for EAGLE
        "data_base_dir": "/data/inspur_disk06/userdir/zhangzhuoming/Quench_Direction_Data/EAGLE/data/",
        "starmass_min": 1.81 * 1e8,
        "redshift": 0.0,
        "starmass_bins": [
            np.log10(1.81 * 1e8 * 10**x) for x in [0, 0.5, 1, 1.5, 2, 10]
        ],
    },
    {
        "name": "SIMBAnofb",
        "snapnum_center": "151",
        "data_base_dir": "/data/inspur_disk06/userdir/zhangzhuoming/Quench_Direction_Data/SIMBAm50n512/nofb/data/",
        "starmass_min": 1.82 * 1e9,
        "redshift": 0.0,
        "starmass_bins": [
            np.log10(1.82 * 1e9 * 10**x) for x in [0, 0.5, 1, 10]
        ],
    },
]

# Global parameters for calculation
part_type_list = ["star_axis"]
axis_align_list = ["all"]
RR200cRatio_bins_list = [np.array([0, 1, 3, 5])]
N_load = 50
sample_angle = 45  # [deg]


# Helper function: Difference calculation
def compute_diff_curves(maj_data, min_data):
    """Computes difference between major and minor axis curves:

    1. Pairwise subtraction: k-th major curve - k-th minor curve
    2. Interpolate minor curve to the major x-grid
    3. Subtract element-wise at identical index i
    4. Force lower/upper bound limits to [0, 1]
    """
    x_maj = maj_data["x"]
    q_maj_list = maj_data["q_list"][1:]
    q_min_list = min_data["q_list"][1:]
    x_min = min_data["x"]

    diff_list = []
    for q_maj, q_min in zip(q_maj_list, q_min_list):
        f_interp = interpolate.interp1d(
            x_maj,
            q_maj,
            kind="linear",
            fill_value="extrapolate",
            bounds_error=False,
        )
        q_maj_interp = f_interp(x_min)
        q_diff = np.clip(q_maj_interp, 0.0, 1.0) - q_min
        diff_list.append(q_diff)

    diff_array = np.array(diff_list)
    return x_min, diff_array


# ===================== 2. Data Processing and Saving Function =====================
def process_and_save_simulation_data(sim_cfg):
    sim_name = sim_cfg["name"]
    snapnum_center = sim_cfg["snapnum_center"]
    data_base_dir = sim_cfg["data_base_dir"]
    starmass_min = sim_cfg["starmass_min"]
    redshift = sim_cfg["redshift"]
    starmass_bins = sim_cfg["starmass_bins"]

    # Output directory for the specific simulation: fig_data_starmass/{sim_name}/
    save_dir = os.path.join("fig_data_starmass", sim_name)
    os.makedirs(save_dir, exist_ok=True)

    RR200cRatio_bins = RR200cRatio_bins_list[0]

    for RR200cRatio_bins_index in range(len(RR200cRatio_bins) - 1):
        # Data collectors for counting and quenched fractions (only star_axis)
        data_collector = {"star_axis": {"major": [], "minor": []}}
        q_frac_collector = {"star_axis": {"major": [], "minor": []}}
        bin_edges = {"star_axis": {"major": None, "minor": None}}
        x_mid_storage = {"star_axis": {"major": None, "minor": None}}

        for axis_align in axis_align_list:
            for part_type in part_type_list:
                dir_path0 = os.path.join("fig_data_R", sim_name + '/' + part_type)
                os.makedirs(dir_path0, exist_ok=True)

                print(
                    f"\n[{sim_name}] Snap: {snapnum_center}, Part: {part_type}, "
                    f"R Bin: ({RR200cRatio_bins[RR200cRatio_bins_index]}, {RR200cRatio_bins[RR200cRatio_bins_index+1]})"
                )

                for N_i in range(N_load + 1):
                    suffix = "ave" if N_i == 0 else f"{N_i-1}"

                    dir_path = os.path.join(
                        "buffer", part_type, str(snapnum_center), axis_align, suffix
                    )
                    if os.path.exists(dir_path):
                        shutil.rmtree(dir_path)
                    os.makedirs(dir_path, exist_ok=True)

                    # Read snapshot galaxy data
                    file_path = os.path.join(
                        data_base_dir,
                        f"GalInHalo_{part_type}",
                        str(snapnum_center),
                        f"galaxy_{suffix}.npy",
                    )
                    if not os.path.exists(file_path):
                        print(f"Warning: File {file_path} not found. Skipping.")
                        continue

                    Full_snap = np.load(file_path)

                    # Mask central galaxies and apply mass cut
                    mask_cgal = Full_snap[:, 8] == Full_snap[:, 9]
                    cgal_data = Full_snap[mask_cgal, :]
                    mask_cgal_M200c = cgal_data[:, 32] >= 1e12
                    cgal_M_data = cgal_data[mask_cgal_M200c, :]

                    # Align state filtering
                    pkl_path = f"align_misalign_data/{snapnum_center}/{axis_align}_40_50_data.pkl"
                    if os.path.exists(pkl_path):
                        with open(pkl_path, "rb") as f:
                            axis_align_data = pickle.load(f)
                        cgal_axis_state = np.array(
                            [sublist[0] for sublist in axis_align_data]
                        )
                        mask_axis_state = np.isin(
                            cgal_M_data[:, 9], cgal_axis_state
                        )
                        cgal_M_data = cgal_M_data[mask_axis_state]

                    # Mask satellite galaxies
                    mask_sgal = (
                        np.isin(Full_snap[:, 9], cgal_M_data[:, 9])
                        & (Full_snap[:, 8] != Full_snap[:, 9])
                        & (Full_snap[:, 7] >= starmass_min)
                    )
                    sgaldata_total = Full_snap[mask_sgal, :]

                    # Radial binning cut
                    sgaldata = []
                    for i in range(len(sgaldata_total)):
                        if (
                            RR200cRatio_bins[RR200cRatio_bins_index]
                            * sgaldata_total[i, 17]
                            < sgaldata_total[i, 3]
                            <= RR200cRatio_bins[RR200cRatio_bins_index + 1]
                            * sgaldata_total[i, 17]
                        ):
                            sgaldata.append(sgaldata_total[i, :])

                    if len(sgaldata) == 0:
                        continue
                    sgaldata = np.vstack(sgaldata)

                    # Filter major axis direction
                    sgaldata_major = []
                    for j in range(sgaldata.shape[0]):
                        dot_products = np.dot(
                            sgaldata[j, 0:3], np.array([1, 0, 0])
                        )
                        angles = np.arccos(
                            np.abs(
                                dot_products
                                / np.linalg.norm(sgaldata[j, 0:3])
                            )
                        )
                        if np.degrees(angles) < sample_angle:
                            sgaldata_major.append(sgaldata[j, :])

                    # Filter minor axis direction
                    sgaldata_minor = []
                    for j in range(sgaldata.shape[0]):
                        dot_products = np.dot(
                            sgaldata[j, 0:3], np.array([0, 0, 1])
                        )
                        angles = np.arccos(
                            np.abs(
                                dot_products
                                / np.linalg.norm(sgaldata[j, 0:3])
                            )
                        )
                        if np.degrees(angles) < sample_angle:
                            sgaldata_minor.append(sgaldata[j, :])

                    # --- Major direction processing ---
                    if len(sgaldata_major) > 0:
                        data = np.vstack(sgaldata_major)
                        m_star, sfr = data[:, 7], data[:, 6]
                        m_SF, _, m_Q = Classify_Galaxies_By_SFR_Mstar(
                            m_star, sfr, redshift=redshift
                        )

                        if bin_edges[part_type]["major"] is None:
                            mid, cnt, bins = Mass_Function(
                                m_star, bins=starmass_bins
                            )
                            bin_edges[part_type]["major"] = bins
                            x_mid_storage[part_type]["major"] = mid
                        else:
                            mid, cnt, _ = Mass_Function(
                                m_star, bins=bin_edges[part_type]["major"]
                            )

                        q_frac = Calc_Quenched_Fraction(
                            m_star, m_Q, bin_edges[part_type]["major"]
                        )
                        data_collector[part_type]["major"].append(cnt)
                        q_frac_collector[part_type]["major"].append(q_frac)

                    # --- Minor direction processing ---
                    if len(sgaldata_minor) > 0:
                        data = np.vstack(sgaldata_minor)
                        m_star, sfr = data[:, 7], data[:, 6]
                        m_SF, _, m_Q = Classify_Galaxies_By_SFR_Mstar(
                            m_star, sfr, redshift=redshift
                        )

                        if bin_edges[part_type]["minor"] is None:
                            mid, cnt, bins = Mass_Function(
                                m_star, bins=starmass_bins
                            )
                            bin_edges[part_type]["minor"] = bins
                            x_mid_storage[part_type]["minor"] = mid
                        else:
                            mid, cnt, _ = Mass_Function(
                                m_star, bins=bin_edges[part_type]["minor"]
                            )

                        q_frac = Calc_Quenched_Fraction(
                            m_star, m_Q, bin_edges[part_type]["minor"]
                        )
                        data_collector[part_type]["minor"].append(cnt)
                        q_frac_collector[part_type]["minor"].append(q_frac)

            # Structure collected data and calculate medians/errors
            data_storage = {"star_axis": {}}
            pt = "star_axis"
            for direc in ["major", "minor"]:
                q_list = q_frac_collector[pt][direc]
                if len(q_list) <= 1:
                    continue

                x_mid = x_mid_storage[pt][direc]
                q_med, q_err = Calc_Stats_For_Errorbar(q_list[1:])

                data_storage[pt][direc] = {
                    "x": x_mid,
                    "q_list": q_list,
                    "q_med": q_med,
                    "q_err": q_err,
                }

            # Compute difference curves and export to NPZ format
            if (
                "major" in data_storage["star_axis"]
                and "minor" in data_storage["star_axis"]
            ):
                # Calculate difference curve for star_axis
                x_star, diff_star_samples = compute_diff_curves(
                    data_storage["star_axis"]["major"],
                    data_storage["star_axis"]["minor"],
                )
                star_med, star_err = Calc_Stats_For_Errorbar(diff_star_samples)

                current_ax3_data = {
                    "x_star": x_star.copy(),
                    "star_median": star_med.copy(),
                    "star_err": star_err.copy(),
                    "RR_low": RR200cRatio_bins[RR200cRatio_bins_index],
                    "RR_high": RR200cRatio_bins[RR200cRatio_bins_index + 1],
                }

                # Save file, e.g., fig_data_starmass/SIMBA/all_R_0_to_1.npz
                save_name = (
                    f"{axis_align_list[0]}_R_{RR200cRatio_bins[RR200cRatio_bins_index]}"
                    f"_to_{RR200cRatio_bins[RR200cRatio_bins_index+1]}.npz"
                )
                save_path = os.path.join(save_dir, save_name)
                np.savez(save_path, **current_ax3_data)
                print(f" Successfully saved data to: {save_path}")


# ===================== 3. Loop Through All Simulations =====================
for sim_cfg in simulations_config:
    print(
        f"\n==================== Starting processing for {sim_cfg['name']} ===================="
    )
    process_and_save_simulation_data(sim_cfg)
```

```python
# 1. Configuration for all four simulations
simulations = [
    {"name": "SIMBA", "title": "SIMBA","xlim": (8, 11), "ylim": (-0.2, 0.4)},
    {"name": "TNG100", "title": "TNG100", "xlim": (8, 11), "ylim": (-0.2, 0.4)},
    {"name": "EAGLE", "title": "EAGLE", "xlim": (8, 11), "ylim": (-0.2, 0.4)},
    {"name": "SIMBAnofb", "title": "SIMBA-nofb", "xlim": (8, 11), "ylim": (-0.2, 0.4)},
]

# Global parameters for plotting
axis_align = "all"
linestyles = ["-", "--", ":"]  # Line styles: 0-1 solid | 1-3 dashed | 3-5 dotted
markers_star = ["o", "o", "o"]
marker_sizes_star = [8, 8, 8]
colors_star = ["#ff7f0e", "#2e8b57", "#1f77b4"]

labels = [
    r"$R/R_\mathrm{200c} \in (\frac{2r_\mathrm{half,c}}{R_\mathrm{200c}},1)$",
    r"$R/R_\mathrm{200c} \in (1,3)$",
    r"$R/R_\mathrm{200c} \in (3,5)$",
]

# Ensure the output figure directory exists
os.makedirs("fig_result", exist_ok=True)


def plot_simulation_quench_fraction(sim_info):
    """Generates and saves the quench fraction plot for a given simulation."""
    sim_name = sim_info["name"]
    title = sim_info["title"]
    xlim = sim_info.get("xlim", (8, 11))
    ylim = sim_info.get("ylim", (-0.2, 0.4))

    # Construct input file paths dynamically for the current simulation
    data_paths = [
        f"fig_data_starmass/{sim_name}/{axis_align}_R_0_to_1.npz",
        f"fig_data_starmass/{sim_name}/{axis_align}_R_1_to_3.npz",
        f"fig_data_starmass/{sim_name}/{axis_align}_R_3_to_5.npz",
    ]

    plt.figure(figsize=(9, 7))

    # Loop through radial bin datasets
    for i, path in enumerate(data_paths):
        if not os.path.exists(path):
            print(
                f"Warning: File not found for {sim_name} at '{path}'. Skipping this curve."
            )
            continue

        dat = np.load(path)

        # Central galaxy properties (star_axis)
        x_s = dat["x_star"]
        med_s = dat["star_median"]
        err_s = dat["star_err"]

        ls = linestyles[i]
        mk_star = markers_star[i]
        msize_star = marker_sizes_star[i]
        c_star = colors_star[i]

        # Plot Central Galaxy errorbar
        plt.errorbar(
            x_s,
            med_s,
            yerr=err_s,
            color=c_star,
            linestyle=ls,
            lw=3,
            marker=mk_star,
            markersize=msize_star,
            markerfacecolor=c_star,
            markeredgecolor="white",
            markeredgewidth=1,
            capsize=4,
            capthick=1.5,
            elinewidth=1.5,
            label=f"{labels[i]}",
        )

    # Figure styling
    starmass_min = next(
    item["starmass_min"]
    for item in simulations_config
    if item["name"] == sim_name
    )
    plt.axhline(0, color="k", linestyle="-", lw=2, alpha=0.7)
    plt.axvspan(
        xmin=8,
        xmax=np.log10(starmass_min),
        color="lightgray",
        alpha=1,
        zorder=0,
    )

    plt.xlabel(r"$\log_{10}(M_* / M_\odot)$", fontsize=20)
    plt.ylabel(r"$\Delta F_\mathrm{Q}$", fontsize=20)

    plt.xticks(fontsize=20)
    plt.yticks(fontsize=20)

    plt.xlim(xlim)
    plt.ylim(ylim)
    plt.grid(alpha=0.3, linestyle="--")
    plt.legend(fontsize=19, loc="upper right")
    plt.title(f"{title}", fontsize=20)
    plt.tight_layout()

    # Save figure to PDF
    output_pdf = f"fig_result/{sim_name}_BCG_all.pdf"
    plt.savefig(output_pdf)
    plt.close()  # Close plot to release memory in loop

    print(f"Successfully generated plot for {sim_name}: {output_pdf}")


# 2. Iterate through all simulations and generate figures
for sim in simulations:
    print(f"Generating plot for simulation: {sim['name']}...")
    plot_simulation_quench_fraction(sim)
```
